```python
#function to extract country data from the HTML
for country in country_cards:
    country_name = country.find(
        class_="country-name").get_text(strip=True)

    capital = country.find(
        class_="country-capital").get_text(strip=True)
    
    population = country.find(
        class_="country-population").get_text(strip=True)

    area = country.find(
        class_="country-area").get_text(strip=True)

    print("Country:", country_name)
    print("Capital:", capital)
    print("Population:", population)
    print("Area:", area)
    print()
```

```python
#changing the list of dictionaries to a pandas dataframe for better analysis
import pandas as pd

countries_df = pd.DataFrame(countries_data)

print(countries_df.head())
print("Dataset size:", countries_df.shape)
```

```python
#checking for missing values
print("\nMissing values:")
print(countries_df.isna().sum())
```

```python
#checking for empty values
print("\nEmpty values:") 
print((countries_df == "").sum())
```

```python
#checking for duplicate values
duplicate_countries = countries_df[
    countries_df["country"].duplicated()
]

print(duplicate_countries)
print("Duplicate country count:", len(duplicate_countries))
```

```python
#convert population and area columns to numeric types
countries_df["population"] = pd.to_numeric(countries_df["population"], errors='coerce')
countries_df["area"] = pd.to_numeric(countries_df["area"], errors='coerce')         
```

```python
#check for failed conversions
failed_population = countries_df[
    countries_df["population"].isna()
]
print(failed_population)
print("Failed. population cnversions:", len(failed_population))

failed_area = countries_df[
    countries_df["area"].isna()]
print(failed_area)
print("Failed area conversions:", len(failed_area))
```

```python
#confirm the transformations
print(countries_df.head())
print("\nUpdated data types:")
print(countries_df.dtypes)
```

```python
#selecting relevant columns for analysis
countries_clean = countries_df[
    [
        "country",
        "capital",
        "population",
      "area"]
].copy()
```

```python
#save the cleaned data to a CSV file
countries_clean.to_csv("countries_clean.csv", index=False)
```
