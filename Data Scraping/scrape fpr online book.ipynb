```python
import requests #send a request to a website and download its page content
import pandas as pd 
from bs4 import BeautifulSoup #helps us search through the page’s HTM
import time #helps pause between requests
```

```python
all_books = []

for page in range(1,51):
    url = f"https://books.toscrape.com/catalogue/page-{page}.html"
    
    response = requests.get(url, timeout=30)
    response.raise_for_status()
    response.encoding = "utf-8"
    
    soup = BeautifulSoup(response.text, "html.parser")
    books = soup.select("article.product_pod")
    
    for book in books:
        all_books.append({
            "title": book.select_one("h3 a")["title"],
            "price": book.select_one(".price_color").get_text(strip=True),
            "availability": book.select_one(
                ".availability").get_text(strip=True),
            "rating": book.select_one(".star-rating")["class"][1]
        })
    print(f"page {page}: scraped {len(books)} books")
    time.sleep(1) #pause between page requests
    
books_df = pd.DataFrame(all_books)

print(books_df.head())
print(f"Total books scraped: {len(books_df)}")

books_df.to_csv("books.csv", index=False)
```
